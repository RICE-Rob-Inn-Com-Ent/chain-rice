```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Sekcja 1: tworzenie danych
np.random.seed(0)
df = pd.DataFrame({
    'x': np.arange(0, 50),
    'y': np.random.randn(50).cumsum()
})

# Sekcja 2: podstawowa analiza
summary = df.describe()
print(summary)

# Sekcja 3: wykres
plt.figure(figsize=(6, 3))
plt.plot(df['x'], df['y'], marker='o')
plt.title('Przykładowy wykres')
plt.xlabel('x')
plt.ylabel('y')
plt.tight_layout()
plt.show()
```
